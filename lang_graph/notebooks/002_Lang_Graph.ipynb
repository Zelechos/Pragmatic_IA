### Imports Libraries

```python
from typing import TypedDict, List
from langgraph.graph import StateGraph
from IPython.display import Image, display
import math
```

### A continuacion creamos el Esquema de nuestro estado

```python
class State(TypedDict):
  name: str
  values: List[int]
  operation: str
```

### Posteriormente Implementamos nuestro nodo

```python
def sum_or_multiple(state: State):
  answer = str(sum(state["values"])) if state["operation"] == "+" else str(math.prod(state["values"]))
  return { "name": "Hi " + state["name"] + ", your answer is : " + answer }
```

### A continuacion estruturamos nuestro grafo

```python
graph = StateGraph(State)

graph.add_node("operation", sum_or_multiple)

graph.set_entry_point("operation")
graph.set_finish_point("operation")

app = graph.compile()

display(Image(app.get_graph().draw_mermaid_png()))
```

### Ahora invocamos y ejecuta nuestra app que contiene el grafo

```python
response = app.invoke({"name": "Jack Sparrow", "values": [1,2,3,4], "operation": "+"})
print(response["name"])
```
